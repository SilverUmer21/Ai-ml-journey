## 1. A function is a contract

A function contract answers:

1. **What does the function do?** — name and docstring.
2. **What enters?** — parameters and their expected types.
3. **What leaves?** — the returned value and expected return type.
4. **What assumptions exist?** — valid ranges or shapes; validation comes in Session 2.

Study the anatomy:

```python
def classify_probability(
    probability: float,
    threshold: float = 0.5,
) -> int:
    """Convert a probability into a binary class label."""
    if probability >= threshold:
        return 1
    return 0
```

| Part | Meaning |
|---|---|
| `classify_probability` | Function name describing the operation. |
| `probability`, `threshold` | Parameters: local names receiving inputs. |
| `: float` | Expected parameter type. |
| `= 0.5` | Default used when the caller omits `threshold`. |
| `-> int` | Expected type of the returned value. |
| Triple-quoted text | Docstring explaining the public contract. |
| `return 1` / `return 0` | Ends the call and sends a value to the caller. |

## 2. `print()` and `return` solve different problems

`print(value)` displays text for a human. It does not send that value back to the caller.

`return value` sends a value to the caller, where it can be stored, tested, or used in another calculation.

A function with no explicit `return` reaches the end and returns `None`.

```python
def display_loss(loss: float) -> None:
    print(f"Loss: {loss}")

def format_loss(loss: float) -> str:
    return f"Loss: {loss}"
```

The first function's useful effect is displaying output, so `-> None` is appropriate. The second returns text, so `-> str` is appropriate.

````python
def display_loss(loss: float) -> None:
    print(f"Loss: {loss}")


def format_loss(loss: float) -> str:
    return f"Loss: {loss}"


displayed_result = display_loss(0.24)
formatted_result = format_loss(0.24)
print("displayed_result:", displayed_result)
print("formatted_result:", formatted_result)
print("usable again:", formatted_result.upper())
````

## 3. Type hints communicate; they do not automatically police

Common annotations:

| Hint | Expected value |
|---|---|
| `int` | Whole number such as `5`. |
| `float` | Decimal number such as `0.72`. |
| `str` | Text. |
| `bool` | `True` or `False`. |
| `list[float]` | List whose items should be floats. |
| `dict[str, int]` | Dictionary with string keys and integer values. |
| `None` | Function intentionally returns no useful value. |

Hints help humans, editors, linters, and type checkers. Ordinary Python usually still runs when the caller provides a different type—until the function body performs an incompatible operation.

## 4. Default parameters

A default makes an argument optional for the caller:

```python
def function(required_parameter: float, optional_parameter: float = 0.5) -> bool:
    ...
```

Required parameters come before parameters with defaults.

- `function(0.7)` uses `0.5`.
- `function(0.7, 0.8)` overrides the default with `0.8`.
- The default is not “forced”; it is only a fallback when omitted.

### Modify with fading guidance

- `probability` is a required `float`.
- `threshold` is a `float` defaulting to `0.5`.
- It returns a `bool`.
- Return whether probability is greater than or equal to threshold.

Unlike the previous task, no exact finished line is shown.

````python
def is_positive_prediction(probability: float, threshold: float = 0.5 ) -> bool:
    """Return whether a probability reaches the classification threshold."""
    if probability >= threshold:
        return True
    return False
````

## 5. Docstrings describe public behavior

A useful function docstring explains what callers need to know—not every implementation line.

For this bootcamp, use this simple style:

```python
def mean_probability(probabilities: list[float]) -> float:
    """Calculate the arithmetic mean of predicted probabilities.

    Args:
        probabilities: Predicted probabilities to average.

    Returns:
        The arithmetic mean as a float.
    """
    return sum(probabilities) / len(probabilities)
```

Notice:

- The first sentence starts with a verb and states the result.
- `Args` describes meaning, not merely repeats the type hint.
- `Returns` explains what the returned value represents.
- No `Raises` section is needed until the function deliberately raises an exception.

````python
def mean_probability(probabilities: list[float]) -> float:
    """Calculate the arithmetic mean of predicted probabilities.

    Args:
        probabilities: Predicted probabilities to average.

    Returns:
        The arithmetic mean as a float.
    """
    return sum(probabilities) / len(probabilities)


print(mean_probability([0.2, 0.5, 0.8]))
help(mean_probability)
````

### Modify the weak docstring

The implementation works, but its docstring is unhelpful. Rewrite only the docstring using the demonstrated structure. Do not change the calculation.

````python
def prediction_range(probabilities: list[float]) -> float:
    """Calculates the range of the given probabilities
       Args: 
            Probabilties to find range of 
       Returns:
            The range as float """
    return max(probabilities) - min(probabilities)
````

````python
# CHECK
assert prediction_range([0.2, 0.5, 0.9]) == 0.7
doc = prediction_range.__doc__ or ""
assert "Args:" in doc, "Add an Args section"
assert "Returns:" in doc, "Add a Returns section"
assert "Does some calculation" not in doc, "Describe the actual purpose"
assert "probabilities" in doc, "Explain the parameter meaning"
print("Passed: the docstring now communicates the contract.")
````

## 6. Make: write one small ML metric

Write this function independently:

```python
calculate_accuracy(y_true: list[int], y_pred: list[int]) -> float
```

Contract:

- `y_true` contains correct binary labels.
- `y_pred` contains predicted binary labels in the same order.
- Count positions where the labels are equal.
- Divide correct predictions by the number of true labels.
- Return a decimal fraction: 3 correct out of 4 returns `0.75`.
- Include a useful `Args` and `Returns` docstring.

For Session 1, assume both lists are non-empty and have equal length. Session 2 will validate those assumptions.

Useful syntax already known:

- `len(values)` gets the number of items.
- `zip(first, second)` pairs items at matching positions.
- `sum(...)` can count Boolean values because `True` behaves like `1`.

Design the steps in English comments before writing the function body.

````python
# MAKE — write your implementation below.
def calculate_accuracy(y_true: list[int] , y_pred: list[int]) -> float:
    """
        Calculates the accuracy of true lables predicted

        Args:
            y_true: contains correct binary labels
            y_pred: contains predicted binary labels

        Returns:
            A accuracy score as float
    """
    count = 0
    true_count = len(y_true)
    for i in range(len(y_true)):
        if y_true[i] == y_pred[i]:
            count+=1 
    return count / true_count
 
````

````python
# AUTOMATED CHECKS
assert calculate_accuracy([1, 0, 1], [1, 0, 1]) == 1.0
assert calculate_accuracy([1, 0, 1, 0], [1, 1, 1, 1]) == 0.5
assert calculate_accuracy([1, 0, 0, 0], [1, 0, 1, 0]) == 0.75
assert calculate_accuracy.__annotations__ == {
    "y_true": list[int],
    "y_pred": list[int],
    "return": float,
}
accuracy_doc = calculate_accuracy.__doc__ or ""
assert "Args:" in accuracy_doc and "Returns:" in accuracy_doc
print("Passed: independently implemented a typed, documented metric.")
````
