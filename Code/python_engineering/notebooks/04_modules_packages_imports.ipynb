## 1. File versus module

### Mental model

A **file** is something stored on disk. A Python **module** is a `.py` file treated by Python as a named unit of code that can be imported.

If a file is named `pricing.py`, its usual module name is `pricing`. The file may define functions, classes, constants, and other names.

### Completed standard-library example

Python already includes a file/module named `math`.

````python
import math

print(math)
print(math.sqrt(81))
print(type(math.sqrt(81)))
````

### Line-by-line trace

| Line | What Python does | What becomes available |
|---|---|---|
| `import math` | Locates and loads the `math` module | The name `math` refers to the module object |
| `math.sqrt` | Looks for the name `sqrt` inside `math` | A function object |
| `math.sqrt(81)` | Calls that function with argument `81` | The returned float `9.0` |

The dot means: look inside the object on the left for the name on the right.

### A second import form

```python
from math import sqrt
result = sqrt(81)
```

This adds `sqrt` directly to the current namespace. It does not add the name `math`. The function still returns a float.

Prefer explicit imports. Avoid `from math import *` because readers cannot easily see where names came from.

## 2. Package and `__init__.py`

### Mental model

A **package** is a directory that groups related Python modules under one name. Your package will be called `ml_utils`.

```text
src/                         package source root
└── ml_utils/                package
    ├── __init__.py          package initialization and public exports

    └── __main__.py          module executed by python -m ml_utils
```

### What `__init__.py` actually does

When Python imports `ml_utils`, it executes `ml_utils/__init__.py`. That file may intentionally expose a small public interface.

Non-ML example:

```text
shop/
├── __init__.py
└── pricing.py
```

```python
# shop/pricing.py
def calculate_total(price: float, quantity: int) -> float:
    return price * quantity

# shop/__init__.py
from shop.pricing import calculate_total

# caller.py
from shop import calculate_total
```

`__init__.py` does not automatically import every module. The author decides what it exposes.

### Import trace

```text
from shop import calculate_total
              │
              ▼
Python locates package shop
              │
              ▼
Python executes shop/__init__.py
              │
              ▼
__init__.py imports calculate_total from shop.pricing
              │
              ▼
Caller receives the name calculate_total
```

### Mental model

Your project uses a `src` layout. The package is not sitting directly beside the terminal. Installation tells the active Python environment where the package lives.

Editable installation does **not** copy your package into a frozen location. It records a link to your working source, so saved source changes are visible without reinstalling each time.

### Exact action H: write `pyproject.toml`

This configuration is new infrastructure, so use the complete scaffold below rather than guessing TOML syntax:

```toml
[build-system]
requires = ["setuptools>=68"]

[project]
name = "ml-utils-learning"
version = "0.1.0"
description = "Small typed ML utilities used for Python engineering practice"
requires-python = ">=3.10"

[tool.setuptools.packages.find]
```

Line meanings:

- `[build-system]` names the tool used to build/install the project.

- `[tool.setuptools.packages.find]` tells setuptools to search inside `src`.

TOML is configuration, not Python. Do not add commas after ordinary key/value lines.

## 8. `__name__`, `__main__.py`, and `python -m`

### Mental model

Every loaded module receives a special variable named `__name__`.

- When a module is **imported**, `__name__` contains its import name, such as `ml_utils.metrics`.
- When a module is the program's **entry point**, `__name__` equals `"__main__"`.

This guard runs code only when the file is the entry point:

```python
if __name__ == "__main__":
    main()
```

### Execution trace

```text
from ml_utils import calculate_accuracy
    └── imported modules receive package/module names
        └── command-line demonstration does not run

python -m ml_utils
    └── Python locates installed package ml_utils
        └── Python executes ml_utils/__main__.py
            └── __name__ equals "__main__"
                └── main() runs
```

`python -m ml_utils` does not mean “run `__init__.py` as the program.” Python executes the package's `__main__.py`. Package initialization may occur as part of loading it.

### Completed non-ML pattern

```python
# weather_tool/__main__.py
from weather_tool.formatting import format_temperature

def main() -> None:
    message = format_temperature(31.5)
    print(message)

if __name__ == "__main__":
    main()
```

`format_temperature` returns a string. `message` stores it. `print` displays it at the outer program boundary.

## 10. Import failures: diagnose instead of hacking

### `ModuleNotFoundError: No module named 'ml_utils'`

Meaning: Python searched its import locations and did not find the package.

Check in this order:

1. Is the correct environment active?
2. Did `python -m pip install -e .` succeed in that environment?
3. Did you run installation from the directory containing `pyproject.toml`?
4. Does `pyproject.toml` tell setuptools to search `src`?
5. Is the directory spelled `ml_utils`?

Do not immediately append paths to `sys.path`. That hides the packaging problem instead of fixing it.

### `ImportError: cannot import name ... from ml_utils`

Meaning: Python found the package, but the requested name was unavailable.

Check spelling, the function/class definition, and the matching export in `__init__.py`. This differs from `ModuleNotFoundError`, where Python could not locate the module/package at all.

### Running an inner file directly

Avoid using this as your package command:

```powershell
python src\ml_utils\metrics.py
```

It changes the import context and encourages demonstration code inside a reusable module. Use package imports and `python -m ml_utils` instead.

### Circular imports

A circular import occurs when module A imports module B while B imports A during initialization. Avoid it here with one direction:

```text
__main__.py ──imports──▶ public package interface
metrics.py  ──imports──▶ validation.py
config.py   ──imports──▶ nothing in this package
validation.py ─imports──▶ nothing in this package
```

Do not make `validation.py` import `metrics.py`.
