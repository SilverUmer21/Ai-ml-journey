# XGBoost Implementation

What this does it : tree 1 -> look at errors -> tree 2 -> look at remaining errors -> tree 3 -> look at remaining errors -> tree 4
So in this way each tree look at previous tree's error and try to do better , this is called "**deliberate practice or boosting in ML terms**"
