```python
import numpy as np 

# we are using 3 neurons for hidden layer which takes 
# A_T as input then computes A_T @ W and 
# then activation function and returns output 
# This forces NumPy to print clean decimals and suppress scientific notation
np.set_printoptions(suppress=True)


def g(z):
    return 1 / (1 + np.exp(-z)) 

def dense(A_T, W, B):
    z = A_T @ W + B
    a_out = g(z)

    return a_out

A_T = np.array([
    [200, 17]
    ])

W = np.array([
    [1, -3, 5],
    [-2, 4, -6]
])

B = np.array([[-1, 1, 2]])

#Manual step by step implementation first then making the code modular
# print(A_T)
# print(W)
# print(B)

# z = A_T @ W + B
# print(f"\nShape of z is : {z.shape} \nz = {z} \n")
# print(f"z1_1 is : {z[0][0]} \n")
# print(f"z1_2 is : {z[0][1]} \n")
# print(f"z1_3 is : {z[0][2]} \n")

layer_1_output = dense(A_T, W, B)

print(f"Layer one activation outputs are: {layer_1_output}")
```
