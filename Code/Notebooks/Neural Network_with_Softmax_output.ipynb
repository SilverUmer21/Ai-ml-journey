```python
##  There are three steps to specifying and  training the model  
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.losses import SparseCategoricalCrossentropy

#  Step 1 is to tell tensorflow how many layers to sequentially string together
model = Sequential(
    [Dense(units = 25 , activation ='relu'),    # first hidden layer 
     Dense(units = 15 , activation = 'relu'),   # second hidden layer 
     Dense(units = 10 , activation = 'softmax') # output layer with softmax as activation function
     ]
)

# Step 2 is to define your loss and cost function 
model.compile(loss = SparseCategoricalCrossentropy)

# Step 3 is to train the Neural Network on dataset X which are the input features 
# and Y which is Ground Truth , it now has multiple outputs
X = None
Y = None
model.fit(X,Y,epochs = 100)
```
